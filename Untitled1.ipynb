```python
import math
import matplotlib.pyplot as plt
import numpy

length = 10
x_coords = [0]*length
for i in range(length):
    x_coords[i] = i+1

y_coords = [0]*length
for i, x in enumerate(x_coords):
    y_coords[i] = 1 - math.log10(x)
    print(x, y_coords[i])
    
plt.figure("Chi", [8,8])

plt.show()
           
```
