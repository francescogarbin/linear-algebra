### What is a matrix?

E' un costrutto che, nel contesto che sto sperimentando, riporta gli n coefficienti di un sistema a n incognite (x, y, z, v, w, etc.) ed il termine noto secondo lo schema:

$
\begin{cases}
a_1 \cdot x + b_1 \cdot y + c_1 \cdot z = d_1 \\
a_2 \cdot x + b_2 \cdot y + c_2 \cdot z = d_2 \\
a_3 \cdot x + b_3 \cdot y + c_3 \cdot z = d_3
\end{cases}
$

Nella forma:

$
\begin{bmatrix}
a_1 & b_1 & c_1 & d_1 \\
a_2 & b_2 & c_2 & d_2 \\
a_3 & b_3 & c_3 & d_3
\end{bmatrix}
$

### Operazioni fondamentali: somma, sottrazione, prodotto

#### Somma
Fattibilità: solo se A e B hanno lo stesso numero di righe e colonne

#### Differenza
Fattibilità: come la somma, solo se A e B hanno lo stesso numero di righe e colonne


#### Prodotto di righe per colonne


#### Dot product $A \cdot B$
Per ogni riga di A, ogni elemento i-esimo della riga di A viene moltiplicato per l'elenmento j-esimo della colonna di B. Se $A_{3\times 2}$ e $B_{4 \times 3}$, l'operazione $A_{2 \times 3} \cdot B_{4 \times 3}$ produce $C_{4 \times 3}$.


Una matrice è una trasformazione.

I sensori della trasformazione sono:

- kernel: l'insieme dei vettori che manda "matrice per vettore a zero" ovvero $A \cdot \vec{v} = 0$ 


- determinante: un valore scalare ottenibile con l'operazione
$
det \begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
= a \cdot d - b \cdot c 
$

Quando la matrice del kernel $\begin{bmatrix} a & b \\ c & d \end{bmatrix}$ ha $det = 0$ e questo porta ad una singola soluzione del sistema (so-called <i>soluzione non banale</i>)

Quando la matrice del kernel ha kernel > 0 e molto grande (quanto "molto grande"?) ottengo infinite soluzioni.

Nel metodo eliminazione di Gauss, se una riga della matrice riporta $\begin{bmatrix}0 & 0 & 0\end{bmatrix}$ significa che ho perso una dimensione a causa di un kernel che è presente perché $A \cdot \vec{v}=0$


#### Relazione tra determinante e kernel

Se $A = \begin{bmatrix} 1 & 1 \\ 2 & 2 \end{bmatrix} $ abbiamo:

- det = 0
- $e_1 \to (1, 2)$
- $e_2 \to (1, 2)$

#### Prodotto di matrici

Ammissibile:

$
\begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{bmatrix} \begin{bmatrix}1 \\ 2 \\ 3 \end{bmatrix}  
$

il risulato è una matrice con una sola colonna.

Ammissibile:
$
\begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{bmatrix}
\begin{bmatrix} 1 & 0 \\ 0 & 2 \\ 3 & 1 \end{bmatrix}  
$

il risultato è una matrice la stessa dimensione di B ovvero 3 righe e 2 colonne.

Ammissibile:

$
\begin{bmatrix}1 & 2 & 3 \end{bmatrix} \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{bmatrix} 
$

Il risultato è una matrice con 3 righe e 3 colonne

Non ammissibile:

$
\begin{bmatrix}1 \\ 2 \\ 3 \end{bmatrix} \begin{bmatrix} 1 & 2 \\ 4 & 5 \\ 7 & 8 \end{bmatrix} 
$

Non ammissibile:

$
AB = \begin{bmatrix}1 \\ 2 \\ 3 \end{bmatrix} \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{bmatrix} 
$



### Cosa ho capito delle Matrici di Eliminazione

Considerando il sistema di equazioni:

$
\left\{
\begin{array}{ll}
    2u + v + w = 5 (1)\\
    4u -6v = -2 (2)\\
    -2u +7v +2w = 9 (3)
\end{array}
\right.
$

Rappresento in forma matriciale:

$
\begin{bmatrix}
     2 &  1 & 1 &  5 \\
     4 & -6 & 0 & -2 \\
    -2 &  7 & 2 &  9 \end{bmatrix}  
$

Procedo con l'eliminazione di Gauss sottraendo 2 volte la (1) alla (2) ottenendo la matrice:

$
\begin{bmatrix}
     2 &  1 & 1 &  5 \\
     0 & -8 & -2 & -12 \\
    -2 &  7 & 2 &  9 \end{bmatrix}  
$

Ragionando sulla matrice come $ Ax = b $, il sistema iniziale sarebbe rappresentabile in <i>forma matriciale</i> come:

$
\begin{bmatrix}
     2 &  1 & 1 \\
     4 & -6 & 0 \\
    -2 &  7 & 2
    \end{bmatrix}  
    \cdot
    \begin{bmatrix}
     u \\
     v \\
     w
    \end{bmatrix}  
    =
    \begin{bmatrix}
      5 \\
     -2 \\
      9 \end{bmatrix}  
$

Posso ottenere la prima eliminazione che ho operato sopra anche moltiplicando una matrice di eliminazione $E_1$ con A e moltiplicandola - separatamente - con b, ovvero $E_1A$ e $E_1b$.

Ecco come: costruisco $E_1$ a partire dalla matrice di identità I e che moltiplico per la matrice A dei coefficienti e la matrice b dei termini noti.

$
I = \begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     0 & 0 & 1 \end{bmatrix}  
$
, 
$
E_1 = \begin{bmatrix}
     1 & 0 & 0 \\
    -2 & 1 & 0 \\
     0 & 0 & 1
    \end{bmatrix}  
$

La matrice di eliminazione riporta la componente $E_{2,1} = -2$ ad indicare che in $Ax = b$ la prima componente due volte alla seconda componente.

Quindi:

$
E_1 \cdot A = 
\begin{bmatrix}
     1 & 0 & 0 \\
    -2 & 1 & 0 \\
     0 & 0 & 1
    \end{bmatrix}
    \cdot
    \begin{bmatrix}
     2 &  1 & 1 \\
     4 & -6 & 0 \\
    -2 &  7 & 2 
    \end{bmatrix}  
    = 
    \begin{bmatrix}
     2  &  1  &  1  \\
     0  & -8  & -2 \\
    -2  &  7  &  2 
    \end{bmatrix}  
$

$
E_1 \cdot b = \begin{bmatrix}
     1 & 0 & 0 \\
    -2 & 1 & 0 \\
     0 & 0 & 1
    \end{bmatrix}
    \cdot
    \begin{bmatrix}
     5 \\
    -2 \\
     9 \end{bmatrix}
     =
     \begin{bmatrix}
     5 \\
    -12 \\
     9 \end{bmatrix}
$

Adesso procedo con una matrice di eliminazione $E_2$, sempre costruita a partire dalla matrice d'identità in cui setto la componente opportuna sulla terza riga della matrice, in questo caso a rappresentare la somma della prima riga alla terza:

$
E_2 = \begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     1 & 0 & 1
    \end{bmatrix}  
$

Quindi $E_2 \cdot A$ produce:

$
E_2 \cdot A = 
\begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     1 & 0 & 1
    \end{bmatrix}
    \cdot
    \begin{bmatrix}
     2  &  1  &  1 \\
     0  & -8  & -2 \\
    -2  &  7  &  2 
    \end{bmatrix}  
    = 
    \begin{bmatrix}
     2  &  1  &  1  \\
     0  & -8  & -2 \\
     0  &  8  &  3 
    \end{bmatrix}  
$

e $ E_2 \cdot b$ produce:

$
E_2 \cdot b = 
\begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     1 & 0 & 1
    \end{bmatrix}
    \cdot
    \begin{bmatrix}
       5 \\
     -12 \\
       9 
    \end{bmatrix}  
    = 
    \begin{bmatrix}
       5 \\
     -12 \\
      14 
    \end{bmatrix}  
$

Quindi il sistema in forma matriciale è diventato:

$
\begin{bmatrix}
     2  &  1  &  1  &  5 \\
     0  & -8  & -2  & -12 \\
     0  &  8  &  3  &  14
     \end{bmatrix}  
$

Posso eliminare la componente ij=(3,2) della matrice sommando la riga 2 alla riga 3, quindi mi invento una bella matrice di sostituzione $E_3$ apposita che somma la riga 3 alla riga 2:

$
E_3 = 
\begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     0 & 1 & 1
    \end{bmatrix}
$

Moltiplico $E_3$ per l'ultima iterazione di A e di b, ovvero:

$
E_3 \cdot A = 
    \begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     0 & 1 & 1
    \end{bmatrix}
    \cdot
    \begin{bmatrix}
     2  &  1  &  1 \\
     0  & -8  & -2 \\
     0  &  8  &  3
     \end{bmatrix}
    =
    \begin{bmatrix}
     2  &  1  &  1 \\
     0  & -8  & -2 \\
     0  &  0  &  1
     \end{bmatrix}
$

e

$
E_3 \cdot b = 
\begin{bmatrix}
     1 & 0 & 0 \\
     0 & 1 & 0 \\
     0 & 1 & 1
    \end{bmatrix}
    \cdot
    \begin{bmatrix}
       5 \\
     -12 \\
      14 
    \end{bmatrix}  
    = 
    \begin{bmatrix}
       5 \\
     -12 \\
       2 
    \end{bmatrix}  
$

Ricomponendo la forma matriciale del sistema di equazioni ottengo:

$
\begin{bmatrix}
     2  &  1  &  1  &  5 \\
     0  & -8  & -2  & -12 \\
     0  &  0  &  1  &  2
     \end{bmatrix}  
$

Da cui capisco ad occhio che $w = 2$ e a questo punto procedo per sostuzione alla soluzione $w = 2, v = 1, u = 1$.
